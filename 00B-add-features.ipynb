```python
import pandas as pd
import numpy as np
import pathlib as pth

from scripts.add_features import add_calendar_features_lgb, add_weather_features_lgb, get_lgb_features_cols
```

```python
def transform_dataframe(path: pth.Path, feature_cols: list[str]) -> pd.DataFrame:
    df = pd.read_csv(path, index_col=[0], parse_dates=["date"])

    df_features = add_calendar_features_lgb(add_weather_features_lgb(df))

    
    df_features_only = df_features[feature_cols]

    if "count" in df.columns:
        df_features_only["count"] = df["count"]

    return df_features_only
```

```python
FEATURE_COLS = get_lgb_features_cols()
```

```python
df_train = transform_dataframe(pth.Path(f"data-00-raw/bike_train.csv"), FEATURE_COLS)

df_train
```

```python
df_test = transform_dataframe(pth.Path(f"data-00-raw/bike_test.csv"), FEATURE_COLS)

df_test
```

```python
df_train.to_csv(f"data-01-processed/lgb_regression_train.csv")

df_test.to_csv(f"data-01-processed/lgb_regression_test.csv")
```
